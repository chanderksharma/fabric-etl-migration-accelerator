### Transformations 1 of 1

Produces `tag_online`, `tag_other`, `stack_channels`, `amount_number`, and 1 more.


```python
tag_online = online_orders.withColumn("SourceStream", F.lit("Online"))
tag_other = other_channels.withColumn("SourceStream", F.lit("Other"))
stack_channels = tag_online.unionByName(tag_other, allowMissingColumns=True)
amount_number = stack_channels.withColumn("AmountNum", F.col("Amount").cast('double'))
amount_by_channel = amount_number.groupBy(*["Region"]).pivot("Channel").agg(F.sum(F.col("Amount").cast('double')))
```
